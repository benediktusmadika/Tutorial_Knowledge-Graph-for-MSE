**Large Language Model (LLM) and Knowledge Graph (KG)**
LLMs consume text prompts, but knowledge graphs convert domain knowledge into explicit entities, relationships, and facts that can be queried, checked, reasoned over, and connected to machine learning.

**1. What is a knowledge graph?**

Intuition:

A knowledge graph is a structured map of things and their relationships.

In materials science, the “things” may be:

materials: NCM811, LiFePO4, MOF-5 \
processes: solid-state synthesis, hydrothermal synthesis, solvothermal synthesis\
properties: capacity, voltage, Gibbs free energy, PLD, LCD\
phases: R-3m, Pnma, Fm-3m\
experiments or simulations: EXP_BAT_001, DFT_CALC_001, MD_RUN_001

Formal definition:

Let

**E={e1,e2,...,en}**

be a set of entities.

Let

**R={r1,r2,...,rm}**

be a set of relation types.

A knowledge graph is a set of facts:

**T⊆E×R×(E∪L)**

where L is a set of literal values such as numbers, strings, temperatures, and units.

Each fact is a triple:

**(s,p,o)**

where:

s = subject
p = predicate / relation
o = object

The W3C RDF data model defines the core structure of RDF as triples consisting of subject, predicate, and object, and says an RDF graph can be visualized as a node-and-directed-arc diagram.

Your guiding sentence is correct:

Knowledge graphs represent entities as nodes and relationships as edges in a graph structure.

Example:

(Material:NCM811, hasFamily, Family:LayeredOxide)

This means:

NCM811 is connected to LayeredOxide by the relation hasFamily.

Graph view:

Material:NCM811 ──hasFamily──> Family:LayeredOxide

**2. Entities**

Intuition:

Entities are the “nouns” of the domain.

In materials science:

NCM811\
LiFePO4\
MOF-5\
solid_state_synthesis\
R-3m\
capacity\
voltage

Formal definition:

An entity is an identifiable object:

$e_i \in E$

Each entity should have:

unique identifier\
type/class\
optional attributes

Example:

Material:NCM811

Here:

Material = entity type\
NCM811 = entity identifier

**3. Relations**

Intuition:

Relations are the “verbs” or “links” between entities.

Examples:

hasFormula\
hasFamily\
usesProcess\
producesPhase\
reportsProperty\
hasTemperatureC

Formal definition:

A relation is a typed connection:

$r_j \in R$
	​

A relation often has a domain and range:

hasFormula:\
    domain = Material\
    range  = Formula

Example:

(Material:NCM811, hasFormula,Formula:LiNi0.8Co0.1Mn0.1O2)


**4. Triples**

A triple is the smallest unit of graph knowledge.

General form:

(subject, predicate, object)

Materials example:

(Experiment:EXP_BAT_001, studiesMaterial, Material:NCM811)

Meaning:

Experiment EXP_BAT_001 studied the material NCM811.

Another example:

(Process:solid_state_750C_O2, hasTemperatureC, TemperatureC:750)

Meaning:

The synthesis process was performed at 750 °C.


**5. Schema and ontology**

A schema tells us which kinds of nodes and relations are allowed.

A simple schema might say:

Experiment --studiesMaterial--> Material\
Material   --hasFormula-------> Formula\
Material   --hasFamily--------> Family\
Experiment --usesProcess------> SynthesisProcess\
Experiment --reportsProperty--> MeasuredProperty

An ontology is a richer schema. It gives formal meaning to classes, relations, constraints, and sometimes rules.

The OWL 2 ontology language provides classes, properties, individuals, and data values; W3C also describes ontologies as formalized vocabularies of terms that define relationships among terms.

For materials science, ontology classes may include:

Material\
CathodeMaterial\
PorousMaterial\
SynthesisProcess\
ThermodynamicProperty\
ElectrochemicalProperty\
StructuralProperty

Ontology relations may include:

isSubclassOf\
hasProperty\
wasSynthesizedBy\
hasCrystalPhase\
hasPoreDescriptor

Example class hierarchy:

LayeredOxideMaterial isSubclassOf CathodeMaterial\
OlivinePhosphateMaterial isSubclassOf CathodeMaterial\
MOF isSubclassOf PorousMaterial



**6. Reasoning**

Reasoning means deriving new facts from existing facts.

Given:

(Material:NCM811, hasFamily, Family:LayeredOxide)\
(Family:LayeredOxide, mapsToClass, Class:CathodeMaterial)

We infer:

(Material:NCM811, rdf:type, Class:CathodeMaterial)

Formal rule:

if (x,hasFamily,LayeredOxide)⇒(x,rdf:type,CathodeMaterial)

Why this matters:

Without reasoning, the graph only stores what was directly observed.

With reasoning, the graph can infer useful scientific categories.


**7. Embeddings**

A graph embedding converts symbolic graph objects into vectors.

Formal definition:

$f(e)→R^d$

where e is an entity and d is the vector dimension.

Example:

Material:NCM811 -> [0.12, -0.44, 0.91, ...]

Why embeddings matter:

Symbolic triples are good for exact reasoning.

Embeddings are useful for machine learning tasks such as:

link prediction
material recommendation
missing property prediction
similar material retrieval
entity disambiguation

In later batches, this connects to PyKEEN-style knowledge graph embedding workflows.

***Python implementation — build the first graph***

We will use:

pandas   -> table representation\
networkx -> graph representation

NetworkX defines a graph as a collection of nodes and identified pairs of nodes called edges, which is enough for our first in-memory knowledge graph.

Code block 1 — raw table → triples → graph

```python
import pandas as pd
import networkx as nx
from collections import Counter

raw_records = [
    {
        "experiment_id": "EXP_BAT_001",
        "material": "NCM811",
        "formula": "LiNi0.8Co0.1Mn0.1O2",
        "family": "LayeredOxide",
        "synthesis_method": "solid_state",
        "temperature_C": 750,
        "atmosphere": "O2",
        "phase": "R-3m",
        "capacity_mAh_g": 196.0,
        "voltage_V": 3.75,
        "deltaG_kJ_mol_Li": -362.0,
        "PLD_A": None,
        "LCD_A": None,
    },
    {
        "experiment_id": "EXP_BAT_002",
        "material": "LiFePO4",
        "formula": "LiFePO4",
        "family": "OlivinePhosphate",
        "synthesis_method": "hydrothermal",
        "temperature_C": 180,
        "atmosphere": "N2",
        "phase": "Pnma",
        "capacity_mAh_g": 158.0,
        "voltage_V": 3.45,
        "deltaG_kJ_mol_Li": -337.0,
        "PLD_A": None,
        "LCD_A": None,
    },
    {
        "experiment_id": "EXP_MOF_001",
        "material": "MOF-5",
        "formula": "Zn4O(BDC)3",
        "family": "MetalOrganicFramework",
        "synthesis_method": "solvothermal",
        "temperature_C": 120,
        "atmosphere": "DMF",
        "phase": "Fm-3m",
        "capacity_mAh_g": None,
        "voltage_V": None,
        "deltaG_kJ_mol_Li": None,
        "PLD_A": 8.0,
        "LCD_A": 12.1,
    },
]

df = pd.DataFrame(raw_records)
#Maps raw table columns to semantic property names and units.
#Example:"capacity_mAh_g": ("DischargeCapacity", "mAh/g")
#means:Column capacity_mAh_g becomes a graph property called DischargeCapacity with unit mAh/g.
property_specs = {
    "capacity_mAh_g": ("DischargeCapacity", "mAh/g"),
    "voltage_V": ("AverageVoltage", "V"),
    "deltaG_kJ_mol_Li": ("GibbsFreeEnergyChange", "kJ/mol_Li"),
    "PLD_A": ("PoreLimitingDiameter", "angstrom"),
    "LCD_A": ("LargestCavityDiameter", "angstrom"),
}
# def node(kind, value):
# Defines a helper function for building node identifiers.
# Converts the value to a string and removes spaces.
# Creates identifiers such as:
# Material:NCM811
# Experiment:EXP_BAT_001
# TemperatureC:750

def node(kind, value):
    clean_value = str(value).replace(" ", "_")
    return f"{kind}:{clean_value}"

triples = []

for row in df.to_dict(orient="records"):   
    # Using orient="records" tells pandas to convert 
    # the DataFrame into a list of dictionaries, where each dictionary represents exactly one row. In each dictionary, the
    # keys are the column names and the values are the data for that row.
    
    exp = node("Experiment", row["experiment_id"])
    mat = node("Material", row["material"])
    formula = node("Formula", row["formula"])
    family = node("Family", row["family"])
    method = node("Method", row["synthesis_method"])
    atmosphere = node("Atmosphere", row["atmosphere"])
    phase = node("Phase", row["phase"])
    # Creates a synthesis-process node that combines method, temperature, and atmosphere.
    # Example: Process:solid_state_750C_O2 
    # 
    process = node(
        "Process",
        f"{row['synthesis_method']}_{row['temperature_C']}C_{row['atmosphere']}",

    )

    # triples.extend([...])
    # Experiment studies Material
    # Material has Formula
    # Material has Family
    # Experiment uses Process
    # Process has Temperature
    # Experiment produces Phase

    
    triples.extend(
        [
            (exp, "rdf:type", "Class:Experiment"),
            (mat, "rdf:type", "Class:Material"),
            (process, "rdf:type", "Class:SynthesisProcess"),
            (exp, "studiesMaterial", mat),
            (mat, "hasFormula", formula),
            (mat, "hasFamily", family),
            (exp, "usesProcess", process),
            (process, "hasMethod", method),
            (process, "hasTemperatureC", node("TemperatureC", row["temperature_C"])),
            (process, "usesAtmosphere", atmosphere),
            (exp, "producesPhase", phase),
        ]
    )

    for column, (property_name, unit) in property_specs.items():
        value = row[column]
        # Skips missing values.
        if pd.notna(value):
            #It creates a unique node for this specific measurement. 
            # By including the experiment_id, it ensures that the "Conductivity" measured in Experiment 
            # A is a distinct node from the "Conductivity" 
            # measured in Experiment B.
            # tes a property node.
            # Example: Property:EXP_BAT_001:DischargeCapacity
            # his is important because property measurements need value and unit.
            prop = node("Property", f"{row['experiment_id']}:{property_name}")
            triples.extend(
                [
                    (exp, "reportsProperty", prop),
                    (prop, "rdf:type", "Class:MeasuredProperty"),
                    (prop, "propertyName", node("PropertyName", property_name)),
                    (prop, "numericValue", node("Value", value)),
                    (prop, "unit", node("Unit", unit)),
                ]
            )

kg = nx.MultiDiGraph()
for subject, predicate, object_ in triples:
    kg.add_node(subject)
    kg.add_node(object_)
    kg.add_edge(subject, object_, relation=predicate)

print("RAW INPUT TABLE")
print(df.to_string(index=False))
print("\nshape:", df.shape)

print("\nINTERMEDIATE TRIPLES")
for i, triple in enumerate(triples[:18], start=1):
    print(f"{i:02d}. {triple}")
print(f"... total triples: {len(triples)}")

print("\nFINAL GRAPH STRUCTURE")
print("nodes:", kg.number_of_nodes())
print("edges:", kg.number_of_edges())
#vCounts how many times each relation type appears.
relation_counts = Counter(data["relation"] for _, _, data in kg.edges(data=True))
for relation, count in sorted(relation_counts.items()):
    print(f"{relation}: {count}")
```

What changed?\
Before:\

After:\

The synthesis process is an entity.\
The phase is an entity.\

The voltage measurement is an entity.\
The unit is explicitly attached.

***Reasoning implementation***

Now we add a simple rule:

LayeredOxide materials are CathodeMaterial.\
OlivinePhosphate materials are CathodeMaterial.\
MetalOrganicFramework materials are PorousMaterial.

```python
rules = {
    "Family:LayeredOxide": "Class:CathodeMaterial",
    "Family:OlivinePhosphate": "Class:CathodeMaterial",
    "Family:MetalOrganicFramework": "Class:PorousMaterial",
}

inferred_triples = []

for subject, object_, data in kg.edges(data=True):
    if data["relation"] == "hasFamily" and object_ in rules:
        inferred_triples.append((subject, "rdf:type", rules[object_]))

for subject, predicate, object_ in inferred_triples:
    kg.add_edge(subject, object_, relation=predicate)

print("INFERRED TRIPLES")
for triple in sorted(inferred_triples):
    print(triple)

print("\nUPDATED GRAPH SUMMARY")
print("nodes:", kg.number_of_nodes())
print("edges:", kg.number_of_edges())

def get_reported_value(experiment, property_name):
    target_property_node = f"PropertyName:{property_name}"
    for _, prop_node, edge_data in kg.out_edges(experiment, data=True):
        if edge_data["relation"] != "reportsProperty":
            continue

        has_name = any(
            data["relation"] == "propertyName" and obj == target_property_node
            for _, obj, data in kg.out_edges(prop_node, data=True)
        )

        if not has_name:
            continue

        value = None
        unit = None
        for _, obj, data in kg.out_edges(prop_node, data=True):
            if data["relation"] == "numericValue":
                value = float(obj.replace("Value:", ""))
            if data["relation"] == "unit":
                unit = obj.replace("Unit:", "")
        return value, unit

    return None, None

print("\nQUERY: cathode experiments with capacity > 180 mAh/g")
for experiment in sorted(n for n in kg.nodes if n.startswith("Experiment:")):
    materials = [
        obj for _, obj, data in kg.out_edges(experiment, data=True)
        if data["relation"] == "studiesMaterial"
    ]
    if not materials:
        continue

    material = materials[0]
    is_cathode = any(
        data["relation"] == "rdf:type" and obj == "Class:CathodeMaterial"
        for _, obj, data in kg.out_edges(material, data=True)
    )
    capacity, cap_unit = get_reported_value(experiment, "DischargeCapacity")
    voltage, volt_unit = get_reported_value(experiment, "AverageVoltage")
    delta_g, dg_unit = get_reported_value(experiment, "GibbsFreeEnergyChange")

    if is_cathode and capacity is not None and capacity > 180:
        print(
            f"{experiment} -> {material}: "
            f"capacity={capacity} {cap_unit}, "
            f"voltage={voltage} {volt_unit}, "
            f"deltaG={delta_g} {dg_unit}"
        )
```

What changed?

Before reasoning:

NCM811 had family LayeredOxide.\
LiFePO4 had family OlivinePhosphate.\
MOF-5 had family MetalOrganicFramework.\

After reasoning:

NCM811 is inferred to be a CathodeMaterial.\
LiFePO4 is inferred to be a CathodeMaterial.\
MOF-5 is inferred to be a PorousMaterial.\

Why it matters:

The raw data never explicitly said:

NCM811 is a cathode material.

The graph derived it from the family rule.

This is one of the first major advantages of knowledge graphs over flat tables.

Exercise 2 — infer material class and query high-voltage cathodes

Task:

Given LiCoO2 as a layered oxide with voltage 3.90 V, infer that it is a cathode material and return it if voltage is at least 3.8 V.

Solution code

```python
import networkx as nx

exercise_triples = [
    ("Experiment:EXP_BAT_003", "studiesMaterial", "Material:LiCoO2"),
    ("Material:LiCoO2", "hasFamily", "Family:LayeredOxide"),
    ("Experiment:EXP_BAT_003", "reportsVoltage", "Voltage:3.90V"),
]

G = nx.MultiDiGraph()
for s, p, o in exercise_triples:
    G.add_edge(s, o, relation=p)

family_rules = {"Family:LayeredOxide": "Class:CathodeMaterial"}

for material, family, edge_data in list(G.edges(data=True)):
    if edge_data["relation"] == "hasFamily":
        inferred_class = family_rules.get(family)
        if inferred_class:
            G.add_edge(material, inferred_class, relation="rdf:type")
            print(f"{material} inferred as {inferred_class}")

print("\nHigh-voltage candidates")
for experiment, voltage_node, edge_data in G.edges(data=True):
    if edge_data["relation"] == "reportsVoltage":
        voltage = float(voltage_node.replace("Voltage:", "").replace("V", ""))
        material = next(
            obj for _, obj, data in G.out_edges(experiment, data=True)
            if data["relation"] == "studiesMaterial"
        )
        is_cathode = any(
            data["relation"] == "rdf:type" and obj == "Class:CathodeMaterial"
            for _, obj, data in G.out_edges(material, data=True)
        )
        if is_cathode and voltage >= 3.8:
            print(f"{experiment} -> {material} at {voltage:.2f} V")
```

What this demonstrates:

Raw fact:
LiCoO2 has family LayeredOxide.

Rule:
LayeredOxide -> CathodeMaterial.

Inferred fact:
LiCoO2 is a CathodeMaterial.

Query:
Find cathode experiments with voltage >= 3.8 V.
